```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
df = pd.read_csv('data/coffee_shop.csv', sep=';')

df
```

The mlxtend library requires data to be in specific format, so we preprocess the data before applying the Apriori algorithm.

True = 1

False = empty values.


Also drop the ID as its not needed for the analysis.

```python
# Drop the ID
df = df.drop(columns='Id')
```

```python
# Replace the NaN values with False
df = df.fillna(False)
```

```python
# Replace 1.0 values with True
df = df.apply(lambda x: x.map(lambda y: True if y == 1.0 else y))
df
```

### Applying the Apriori Algorithm

Now the data is preprocessed and ready.


```python
frequent_itemsets = apriori(df, min_support=0.1, use_colnames=True)
frequent_itemsets
```

```python
# Generate association rules
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.5)

# Sort in descending order of confidence
rules = rules.sort_values(by='confidence', ascending=False)

rules
```

```python
rules = association_rules(frequent_itemsets, metric='lift', min_threshold=2)

# Sort in descending order of lift
rules = rules.sort_values(by='lift', ascending=False)
rules
```
