### Clustering the Data

There are a number of ways to identify the optimum number of clusters. The method below is known as the Elbow method.

The Elbow plot allows to plot the inertia, which is a measure of how well the data was clustered by the K-Means algorithm, agaisnt the number of clusters. From this plot, we are looking for a point where the inertia begins to slow.

```python
# Function to work out optimum number of clusters.

def optimise_k_means(data, max_k):
    means = []
    inertias = []

    for k in range(1, max_k):
        kmeans = KMeans(n_clusters=k)
        kmeans.fit(data)

        means.append(k)
        inertias.append(kmeans.inertia_)

    # Generate the Elbow plot
    fig=plt.subplots(figsize=(10, 5))
    plt.plot(means, inertias, 'o-')
    plt.xlabel('Number of Clusters')
    plt.ylabel('Inertia')
    plt.grid(True)
    plt.show()
```

```python
optimise_k_means(df[['RHOB_T', 'NPHI_T']], 10)
```

### Creating Multiple Clusters

```python
for k in range(1, 6):
    kmeans=KMeans(n_clusters=k)
    kmeans.fit(df[['RHOB_T', 'NPHI_T']])
    df[f'KMeans_{k}'] = kmeans.labels_
```
